## Instalação do Python

Para começar a programar em Python, precisamos instalar o interpretador e configurar o ambiente de desenvolvimento.

### 1. Baixe o Python

Acesse o site oficial:

[Download do Python](https://www.python.org/downloads/?utm_source=chatgpt.com)

Escolha a versão mais recente e estável disponível para o seu sistema operacional.

---

### 2. Instalação no Windows

Durante a instalação, marque a opção:

> **Add Python.exe to PATH**

Depois, clique em **Install Now**.

Essa opção é importante porque permite executar o Python diretamente pelo terminal.

Para verificar se a instalação foi concluída, abra o **Prompt de Comando** ou o **PowerShell** e execute:

```bash
python --version
```

Ou:

```bash
py --version
```

Se tudo estiver correto, será exibida a versão instalada, por exemplo:

```text
Python 3.x.x
```

---

### 3. Instalação no Linux

Em distribuições baseadas em Debian, Ubuntu ou Linux Mint, normalmente o Python já vem instalado. Para verificar:

```bash
python3 --version
```

Caso seja necessário instalar:

```bash
sudo apt update
sudo apt install python3 python3-pip
```

O `pip` é o gerenciador de pacotes do Python e permite instalar bibliotecas externas.

---

### 4. Instalação do VS Code

Para desenvolver em Python, uma excelente opção é o **Visual Studio Code**.

[Visual Studio Code](https://code.visualstudio.com/?utm_source=chatgpt.com)

Após instalar o VS Code, procure pela extensão:

> **Python — Microsoft**

Ela oferece recursos como:

* Autocompletar código;
* Depuração;
* Verificação de erros;
* Suporte a ambientes virtuais;
* Execução de scripts Python;
* Integração com notebooks Jupyter.

---

### 5. Primeiro programa

Crie um arquivo chamado:

```text
hello.py
```

Adicione:

```python
print("Olá, mundo!")
```

Execute pelo terminal:

```bash
python hello.py
```

No Linux, pode ser necessário usar:

```bash
python3 hello.py
```

Resultado:

```text
Olá, mundo!
```

### Exercício prático

Instale o Python, configure o VS Code e crie um programa que exiba:

```text
Olá! Estou começando a aprender Python.
```

Depois, você pode avançar para **variáveis e tipos de dados**, que são os próximos fundamentos da linguagem.
