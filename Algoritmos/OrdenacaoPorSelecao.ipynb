* **Construção da nova lista / Remoção:**
* No modelo teórico trazido pelo livro, o elemento encontrado é removido do vetor/lista original e anexado em uma nova lista encadeada ou vetor [[07:37](http://www.youtube.com/watch?v=CGZpwC69Sx0&t=457), [08:28](http://www.youtube.com/watch?v=CGZpwC69Sx0&t=508)].

## ⏱️ 2. Análise de Desempenho e Complexidade (Notação Big O)

* **Complexidade Quadrática $O(N^2)$:**
* Para encontrar o menor ou maior elemento em uma lista de $N$ itens, é preciso fazer uma varredura linear de tempo $O(N)$ [[10:16](http://www.youtube.com/watch?v=CGZpwC69Sx0&t=616)].
* Como esse processo de varredura precisa ser repetido $N$ vezes (uma para cada item), o tempo de execução total é $O(N \times N) = O(N^2)$ [[10:42](http://www.youtube.com/watch?v=CGZpwC69Sx0&t=642), [11:03](http://www.youtube.com/watch?v=CGZpwC69Sx0&t=663)].


* **Ajuste Médio e Constantes:**
* À medida que os elementos vão sendo removidos, o tamanho da lista diminui ($N$, $N-1$, $N-2$, ...), resultando em uma média de $\frac{1}{2} \times N$ elementos analisados por ciclo [[13:19](http://www.youtube.com/watch?v=CGZpwC69Sx0&t=799), [13:39](http://www.youtube.com/watch?v=CGZpwC69Sx0&t=819)].
* Na Notação Big O, fatores constantes (como o $\frac{1}{2}$) são ignorados na análise do pior caso, mantendo o algoritmo classificado como **$O(N^2)$** [[13:53](http://www.youtube.com/watch?v=CGZpwC69Sx0&t=833), [14:33](http://www.youtube.com/watch?v=CGZpwC69Sx0&t=873)].


* **Desempenho Geral:** O autor reforça que o *Selection Sort* é um algoritmo simples de compreender, porém relativamente lento em comparação com algoritmos como o *Quicksort* ($O(N \log N)$) [[01:49](http://www.youtube.com/watch?v=CGZpwC69Sx0&t=109), [15:18](http://www.youtube.com/watch?v=CGZpwC69Sx0&t=918)].

# Complexidade do algoritmo
A razão pela qual o algoritmo de **Ordenação por Seleção (*Selection Sort*)** possui complexidade de tempo **$O(N^2)$** está na combinação de **dois loops encadeados** (um loop dentro do outro) e na matemática por trás de quantas verificações ele precisa fazer.

---

### 1. A Intuição do Algoritmo

Imagine que você tem uma lista desordenada com $N$ cartas numeradas. Para ordená-las com o *Selection Sort*:

1. **Passo 1:** Você percorre a lista inteira para encontrar a menor carta e a coloca na 1ª posição.
2. **Passo 2:** Você percorre o restante da lista para achar a segunda menor carta e a coloca na 2ª posição.
3. **Passo 3:** Repete o processo até que a última carta esteja no lugar correto.

Note que existem dois níveis de trabalho acontecendo:

* **Loop Externo:** Quantas vezes você precisa repetir o processo? **$N$ vezes** (uma para cada elemento da lista).
* **Loop Interno:** Em cada rodada, quantas cartas você precisa olhar para achar a menor? **Aproximadamente $N$ cartas**.

Se você faz um trabalho de ordem $N$, $N$ vezes seguidas, o resultado é de ordem $N \times N = N^2$.

---

### 2. A Matemática Exata das Comparações

Na prática, a cada rodada a lista diminui porque as primeiras cartas já estão organizadas. Vamos somar quantas verificações são feitas em cada passo para uma lista de tamanho $N$:

* Na **1ª rodada:** você verifica $N$ elementos.
* Na **2ª rodada:** você verifica $N - 1$ elementos.
* Na **3ª rodada:** você verifica $N - 2$ elementos.
* ...
* Na **última rodada:** você verifica apenas $1$ elemento.

O total de operações é a soma dessa sequência decrecente:


$$\text{Total de operações} = N + (N - 1) + (N - 2) + \dots + 2 + 1$$

Essa é uma **Progressão Aritmética (PA)** bem conhecida na matemática. A fórmula para calcular o resultado dessa soma é:

$$\text{Soma} = \frac{N \times (N + 1)}{2} = \frac{N^2 + N}{2} = \frac{1}{2}N^2 + \frac{1}{2}N$$

---

### 3. Por que isso vira $O(N^2)$ no Big O?

Na análise de algoritmos com a Notação Big O, nós nos importamos com o **comportamento no pior cenário quando $N$ cresce para números gigantescos** (como $1.000.000$ de itens). Por isso, aplicamos duas regras de simplificação:

1. **Ignoramos as constantes multiplicativas:** O termo $\frac{1}{2}$ é apenas uma constante. Não importa se um computador é 2 vezes mais rápido que outro; a taxa de crescimento da curva continua a mesma.
2. **Mantemos apenas o termo dominante (de maior expoente):** Comparando os termos $\frac{1}{2}N^2$ e $\frac{1}{2}N$:
* Se $N = 100$: $N^2 = 10.000$ (o termo $N^2$ domina $99\%$ do resultado).
* Se $N = 1.000.000$: $N^2 = 1.000.000.000.000$ (o termo $N$ torna-se insignificante perto de $N^2$).



Ao remover os coeficientes e os termos menores, a expressão $\frac{1}{2}N^2 + \frac{1}{2}N$ reduz-se a:

$$\mathbf{O(N^2)}$$

---

### 💡 Em resumo

O *Selection Sort* é $O(N^2)$ porque ele **não tira proveito do que já foi ordenado anteriormente**. Ele é "cego" em relação à estrutura da lista: mesmo que a lista já esteja quase toda ordenada, ele obrigatoriamente fará todas as varreduras de busca pelo menor elemento até o fim, exigindo uma quantidade quadrática de operações.

```python
def buscaMenor(arr):
    menor = arr[0]
    menor_indice = 0

    for i in range(1, len(arr)):
        if arr[i] < menor:
            menor = arr[i]
            menor_indice = i
    return menor_indice

def ordenacaoporSelecao(arr):
    novoArr = []

    for i in range(len(arr)):
        menor = buscaMenor(arr)
        novoArr.append(arr.pop(menor))

    return novoArr

print(ordenacaoporSelecao([5, 3, 6, 2, 10]))
```
